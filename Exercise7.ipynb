```python
s1={1,2,3,4}
s2={3,4,7,8,9}
print(s1|s2)
print(s1&s2)
print(s1-s2)
print(s1^s2)
```

```python
s1={1,2,3,4}
s2={3,4,7,8,9}
s1.add(10)
print(s1)
s1.update(((10,20,30)))
print(s1)
s1.remove(10)
print(s1)
s1.discard((20,30))
print(s1)
s1.pop()
print(s1)
s2.clear()
print(s2)
print(s1.union(s2))
print(s1.intersection(s2))
print(s1.difference(s2))
```
