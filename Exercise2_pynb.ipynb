```python
#3.Example for match case
day=int(input("Enter the first number "))
match day:
  case 1 :
    print("sunday")
  case 2:
    print("Monday")
  case 3:
    print("Tuesday")
  case 4:
    print("Wednesday")
  case 5:
    print("Thuesday")
  case 6 :
    print("Friday")
  case 7:
    print("Saturday")
  case _:
    print("Invalid input")
```
