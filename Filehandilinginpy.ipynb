```python
#program on w mode
file=open("f.txt","w")
file.write("this is the first file you have created")
print("File created")
file.close()
```

```python
#program on r mode
with open("f.txt","r") as f:
  data=f.read()
  print(data)
```

```python
with open("f.txt","a") as f:
  f.writelines(["\napple","\nname","\nend"])

```

```python
with open("f.txt","r") as f:
  print(f.readline())
```

```python
with open("f.txt","r") as f:
  print(f.readlines())
```

```python
with open("f.txt","r") as f:
  print(f.readline())
  print(f.tell())
  f.seek(0)
  print(f.tell())
  print(f.readline())
```

```python
#program to copy content of one file to another file
file=open("f.txt","r")
file2=open("f1.txt","w")
for line in file:
  file2.write(line)
file.close()
file2.close()
```

```python
#read each line in file in reverse mode
with open("f.txt","r") as f:
  for line in f:
    print(line[::-1])
```

```python
file=open("f.txt","r")
file3=open("f3.txt","w")
for line in file :
  file3.write(line)
file.close()
file3.close()
```

```python
with open("f.txt", "r") as file:
    lines = file.readlines()
lines = [line.lower() for line in lines]
lines.sort()
lines = [line.lower() for line in lines]
lines.sort()
with open("output.txt", "w") as file:
    file.writelines(lines)
```

```python
import pickle
with open("s1.dat","wb") as f:
  data={"name":"bhargav","age":20,"city":"usa"}
  pickle.dump(data,f)
with open("s1.dat","rb") as f2:
  data=pickle.load(f2)
  print(data)
```
